```python
import exifread

with open('/alpha/lobarev_e/Cube++/auxiliary/source/CR2/00_0010.CR2', 'rb') as f:
    tags = exifread.process_file(f)
    
    # Print all EXIF tags
    for tag in tags.keys():
        if tag not in ('JPEGThumbnail', 'TIFFThumbnail', 'Filename', 'EXIF MakerNote'):
            print(f"{tag:30}: {tags[tag]}")
    
    # Example: access specific values
    print("Camera Model:", tags.get('Image Model'))
    print("Exposure Time:", tags.get('EXIF ExposureTime'))
    print("Focal Length:", tags.get('EXIF FocalLength'))
```
